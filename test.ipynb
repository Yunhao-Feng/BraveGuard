```python
import json

REQUIRED_KEYS = {
    "id",
    "category",
    "jailbreak_method",
    "query",
    "decomposed_query",
    "comment",
}

file_path = "data/subset.json"
report_path = "data/subset_missing_report.json"

with open(file_path, "r", encoding="utf-8") as f:
    data = json.load(f)

report = []

for idx, item in enumerate(data):
    if not isinstance(item, dict):
        report.append({
            "index": idx,
            "missing_keys": list(REQUIRED_KEYS),
            "reason": f"item is not an object: {type(item).__name__}"
        })
        continue

    missing = sorted(REQUIRED_KEYS - set(item.keys()))
    if missing:
        report.append({
            "index": idx,
            "missing_keys": missing,
            "item": item
        })

with open(report_path, "w", encoding="utf-8") as f:
    json.dump(report, f, ensure_ascii=False, indent=2)

print(f"检查完成，发现 {len(report)} 个异常项，已写入 {report_path}")
```

```python
import os
import shutil
import glob

# 定义源目录和目标目录
src_dir = "exports_local"
dst_dir = "exports"

# 确保目标目录存在
os.makedirs(dst_dir, exist_ok=True)

# 查找所有 jsonl 文件
jsonl_files = glob.glob(os.path.join(src_dir, "*.jsonl"))

if not jsonl_files:
    print(f"⚠️  {src_dir} 下没有找到任何 jsonl 文件")
else:
    # 移动文件
    moved_count = 0
    for file_path in jsonl_files:
        file_name = os.path.basename(file_path)
        dst_path = os.path.join(dst_dir, file_name)
        shutil.move(file_path, dst_path)
        print(f"✅ 已移动: {file_name}")
        moved_count += 1

    print(f"\n📦 本次共移动了 {moved_count} 个 jsonl 文件")

# 统计 exports 下所有 jsonl 文件
total = len(glob.glob(os.path.join(dst_dir, "*.jsonl")))
print(f"📊 exports 目录下现共有 {total} 个 jsonl 文件")

```
